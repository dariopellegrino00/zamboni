# The Zamboni environment

The initial state of an episode, drawn by `ZamboniEnv.render()`: the rink, the target with its 2 m success disk (red) and the robot at rest at the center (blue). The gray margin is outside the rink.

```python
import matplotlib.pyplot as plt

from zamboni.env import ZamboniEnv

env = ZamboniEnv(render_mode="rgb_array")
obs, info = env.reset(seed=0)
obs_rounded = obs.round(2)
print(
    f"state [x: {obs_rounded[0]}, "
    f"y: {obs_rounded[1]}, "
    f"vx: {obs_rounded[2]}, "
    f"vy: {obs_rounded[3]}, "
    f"target_x: {obs_rounded[4]}, "
    f"target_y: {obs_rounded[5]}, "
    f"steps_left: {obs_rounded[6]}]",
)

plt.figure(figsize=(8, 5))
plt.imshow(env.render())
plt.axis("off")
plt.show()
```

Each seed gives a different target, uniform on the rink and at least 2 m from the robot.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 3.5))
for ax, seed in zip(axes, (1, 2, 3), strict=True):
    obs, _ = env.reset(seed=seed)
    ax.imshow(env.render())
    ax.set_title(f"seed {seed}: target at ({obs[4]:.1f}, {obs[5]:.1f}) m")
    ax.axis("off")
plt.show()
```
